# **INTRODUCTION**

Diabetes is a metabolic disease that causes high blood sugar. The hormone insulin moves sugar from the blood into your cells for storage or energy. With diabetes, your body either doesn't make enough insulin or can't effectively use the insulin it does make.

Untreated high blood sugar from diabetes can damage your nerves, eyes, kidneys, and other organs. Understanding the different types of diabetes and their risk factors is crucial for early detection and management. This risk predictor aims to identify individuals who may be at a higher risk of developing diabetes based on various health indicators.

# 1. Import Necessary Libraries

We begin by importing the Python libraries essential for data manipulation, machine learning, and visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report
```

# 2. Load the Dataset

For this project, we'll assume a dataset is available (e.g., 'diabetes.csv'). You would typically upload this file to your Colab environment or load it directly from a source like Google Drive or a public URL. This dataset usually contains various health metrics and a target variable indicating diabetes presence.

```python
# To upload a file, you can use the following code:
# from google.colab import files
# uploaded = files.upload()
# For this example, let's assume 'diabetes.csv' is in the current directory or uploaded.

# Load the dataset
try:
    df = pd.read_csv('diabetes.csv')
    print("Dataset loaded successfully.")
except FileNotFoundError:
    print("Error: 'diabetes.csv' not found. Please upload the file or provide the correct path.")
    # Create a dummy DataFrame for demonstration if file is not found
    data = {
        'Pregnancies': [6, 1, 8, 1, 0],
        'Glucose': [148, 85, 183, 89, 137],
        'BloodPressure': [72, 66, 64, 66, 40],
        'SkinThickness': [35, 29, 0, 23, 35],
        'Insulin': [0, 0, 0, 94, 168],
        'BMI': [33.6, 26.6, 23.3, 28.1, 43.1],
        'DiabetesPedigreeFunction': [0.627, 0.351, 0.672, 0.167, 2.288],
        'Age': [50, 31, 32, 21, 33],
        'Outcome': [1, 0, 1, 0, 1]
    }
    df = pd.DataFrame(data)
    print("Using a dummy DataFrame for demonstration.")

# Display the first 5 rows of the dataset
display(df.head())
```

# 3. Exploratory Data Analysis (EDA)

EDA is crucial for understanding the dataset's characteristics, identifying patterns, and detecting anomalies. We'll check for basic information, summary statistics, and visualize distributions and correlations.

```python
print("\nDataset Information:")
df.info()

print("\nDescriptive Statistics:")
display(df.describe())

print("\nNumber of missing values per column:")
display(df.isnull().sum())

# Check for zero values in columns where it's not logical (e.g., Glucose, BloodPressure, BMI)
print("\nNumber of zero values in key columns:")
for column in ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']:
    print(f"{column}: {(df[column] == 0).sum()}")
```

### Visualizing Data Distributions

Histograms help us understand the distribution of each feature. We'll also visualize the distribution of the target variable (`Outcome`).

```python
df.hist(figsize=(15, 10), bins=20)
plt.suptitle('Distribution of Features')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

sns.countplot(x='Outcome', data=df)
plt.title('Distribution of Diabetes Outcome (0: No Diabetes, 1: Diabetes)')
plt.show()
```

### Correlation Heatmap

A heatmap helps visualize the correlation between different features. This can highlight highly correlated features that might impact model performance.

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Features')
plt.show()
```

# 4. Data Preprocessing

This phase involves cleaning and transforming the raw data into a suitable format for machine learning. This typically includes handling missing values, scaling numerical features, and splitting the dataset.

### Handling Zero Values

In some datasets, a value of `0` in columns like `Glucose`, `BloodPressure`, `SkinThickness`, `Insulin`, and `BMI` might represent missing data, as these values cannot logically be zero for a living person. We will replace these zeros with the mean of their respective columns for a more robust model. 'Pregnancies' can logically be zero, so it's excluded.

```python
for col in ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']:
    df[col] = df[col].replace(0, df[col].mean())

print("\nZero values replaced with column means:")
for column in ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']:
    print(f"{column}: {(df[column] == 0).sum()}") # Should all be 0 now

display(df.describe()) # Check updated statistics
```

### Splitting Features and Target Variable

We separate the dataset into features (X) and the target variable (y).

```python
X = df.drop('Outcome', axis=1) # Features
y = df['Outcome']             # Target variable

print(f"Features (X) shape: {X.shape}")
print(f"Target (y) shape: {y.shape}")
```

### Splitting Data into Training and Testing Sets

We divide the data into training and testing sets to evaluate the model's performance on unseen data. A common split is 80% for training and 20% for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42, stratify=y)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

### Feature Scaling

Scaling numerical features to a standard range (e.g., mean=0, variance=1) is crucial for many machine learning algorithms, especially those that rely on distance calculations, to prevent features with larger values from dominating the learning process.

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Features scaled successfully.")
display(pd.DataFrame(X_train_scaled, columns=X.columns).head())
```

# 5. Model Training

We will train a classification model to predict diabetes risk. Logistic Regression is a good baseline model, and RandomForest Classifier is a powerful ensemble method often used for its robustness and accuracy.

### Logistic Regression Model

```python
log_reg_model = LogisticRegression(random_state=42)
log_reg_model.fit(X_train_scaled, y_train)

print("Logistic Regression Model trained successfully.")
```

### Random Forest Classifier Model

```python
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train_scaled, y_train)

print("Random Forest Classifier Model trained successfully.")
```

# 6. Model Evaluation

After training, we evaluate the models' performance on the test set using various metrics such as accuracy, precision, recall, F1-score, and the ROC AUC score. A confusion matrix provides a detailed breakdown of correct and incorrect predictions.

### Evaluate Logistic Regression

```python
y_pred_lr = log_reg_model.predict(X_test_scaled)
y_proba_lr = log_reg_model.predict_proba(X_test_scaled)[:, 1]

print("Logistic Regression Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_lr):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_lr):.4f}")
print(f"Recall: {recall_score(y_test, y_pred_lr):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred_lr):.4f}")
print(f"ROC AUC Score: {roc_auc_score(y_test, y_proba_lr):.4f}")

print("\nConfusion Matrix:")
sns.heatmap(confusion_matrix(y_test, y_pred_lr), annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['No Diabetes', 'Diabetes'], yticklabels=['No Diabetes', 'Diabetes'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Logistic Regression Confusion Matrix')
plt.show()

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))
```

### Evaluate Random Forest Classifier

```python
y_pred_rf = rf_model.predict(X_test_scaled)
y_proba_rf = rf_model.predict_proba(X_test_scaled)[:, 1]

print("Random Forest Classifier Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_rf):.4f}")
print(f"Recall: {recall_score(y_test, y_pred_rf):.4f}")
print(f"F1-Score: {f1_score(y_test, y_pred_rf):.4f}")
print(f"ROC AUC Score: {roc_auc_score(y_test, y_proba_rf):.4f}")

print("\nConfusion Matrix:")
sns.heatmap(confusion_matrix(y_test, y_pred_rf), annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['No Diabetes', 'Diabetes'], yticklabels=['No Diabetes', 'Diabetes'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Random Forest Confusion Matrix')
plt.show()

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```
